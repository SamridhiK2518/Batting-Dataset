**Data Dictionary**
1. RunsScored : Number of Runs scored by Kohli in the match.
2. BallsFaced : Number of Balls Faces by Kohli in the match.
3. BattingPosition : Batting position is Virat Kohli's position on the batting order.
4. Dismissal : How Kohli got out in the match or not out. Caught, NO: Not Out, Bowled, Run out, LBW, Stumped, Hit wicket.
5. Inning : An innings is one of the divisions of a match during which one team takes its turn to bat. Two 1st inning or 2nd inning.
6. Opposition : Name of the Opposition team.
7. Year : Year in which the match was played.

## Q-1 What is the percentage for Kohli being Not Out.
**Make a Pie Chart displaying the distribution of Dismissals**

```python
mydata["Dismissal"].value_counts()
not_out_count = (mydata["Dismissal"] == "NO").sum()

total_innings = len(mydata)

not_out_percentage = (not_out_count / total_innings) * 100

print("Percentage of Kohli being Not Out:", not_out_percentage)
dismissal_counts = mydata["Dismissal"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    dismissal_counts,
    labels=dismissal_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Virat Kohli's Dismissals")
plt.show()
```

## Q-2 Find the number of times when kohli scored less than or equal to 7 runs?

```python
count = (mydata["RunsScored"] <= 7).sum()

print("Number of times Kohli scored <= 7 runs:", count)
```

## Q.4 Use Boxplot to tell the IQR( Inter Quartile Range) middle 50% of runs scored by Kohli?
[Hint: Middle 50% Range= Q3(75%)-Q1(25%)]

```python
# Boxplot of Runs Scored

plt.figure(figsize=(8, 5))

sns.boxplot(x=mydata["RunsScored"])

plt.title("Boxplot of Runs Scored by Virat Kohli")
plt.xlabel("Runs Scored")

plt.show()
Q1 = mydata["RunsScored"].quantile(0.25)
Q3 = mydata["RunsScored"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 (25%):", Q1)
print("Q3 (75%):", Q3)
print("IQR:", IQR)
```

## Q. Find out some interesting dismissal patterns in Kohli's career againsts the oppositions.
[Hint: Stacked Bar Plot]

```python
# Create a table of Opposition vs Dismissal
dismissal_pattern = pd.crosstab(
    mydata["Opposition"],
    mydata["Dismissal"]
)

# Stacked bar plot
dismissal_pattern.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title("Dismissal Patterns Against Different Oppositions")
plt.xlabel("Opposition")
plt.ylabel("Number of Dismissals")

plt.xticks(rotation=45)
plt.legend(title="Dismissal")
plt.show()
```

## Q. Which opposition has Kohli scored most runs against? Find the median score against different oppositions and use a boxplot to represent it?
[Hint: Use barchart & BoxPlot]

```python
median_scores = mydata.groupby("Opposition")["RunsScored"].median()

print(median_scores)
```

```python
plt.figure(figsize=(12, 6))

median_scores.plot(kind="bar")

plt.title("Median Runs Scored Against Each Opposition")
plt.xlabel("Opposition")
plt.ylabel("Median Runs")

plt.xticks(rotation=45)
plt.show()
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=mydata,
    x="Opposition",
    y="RunsScored"
)

plt.title("Runs Scored Against Different Oppositions")
plt.xlabel("Opposition")
plt.ylabel("Runs Scored")

plt.xticks(rotation=45)
plt.show()
```

Adding a column giving true for those rows which represent post transformation 

```python
mydata['post_transform'] = mydata["Year"]>2012
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.boxplot(
    x='post_transform',
    y='RunsScored',
    data=mydata
)

plt.xticks([0, 1], ['Pre-transformation', 'Post-transformation'])
plt.xlabel('Transformation')
plt.ylabel('Runs Scored')
plt.title('Kohli: Pre vs Post Transformation')

plt.show()
```

## Can we say there is a  linear relationship between RunsScored & BallsFaced over different batting positions?
[Hint: Use LMplot]

```python
sns.lmplot(
    data=mydata,
    x='BallsFaced',
    y='RunsScored',
    hue='BattingPosition',
    height=6,
    aspect=1.5
)

plt.xlabel('Balls Faced')
plt.ylabel('Runs Scored')
plt.title('Relationship Between Runs Scored and Balls Faced by Batting Position')

plt.show()
```

## How the batting position has changed with different oppositions yearly?
[Hint: FacetGrid]

```python
import seaborn as sns
import matplotlib.pyplot as plt

g = sns.FacetGrid(
    mydata,
    col='Opposition',
    col_wrap=4,
    height=4,
    sharex=False,
    sharey=True
)

g.map_dataframe(
    sns.lineplot,
    x='Year',
    y='BattingPosition',
    marker='o'
)

g.set_axis_labels('Year', 'Batting Position')
g.set_titles('{col_name}')
g.fig.suptitle(
    'Yearly Change in Batting Position Against Different Oppositions',
    y=1.02
)

plt.show()
```
